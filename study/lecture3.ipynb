## Focus for Today
**Statistical Bias**: the difference between you estimate and the truth

Here are the three concepts of sampling:

* Population: The set of all units of *interest*, size $N$
* Sampling frame: the set of all possible units that can be drawn into the sample
* Sample: a subset of the sampling frame, size $n$.

### Scenario 1
A census

Some key features of a census are:
* population <-> sampling frame <-> sample
* Pros: Lots of data!
    * No selection bias
    * Easy inference
* Cons: Expensive (time, money, collecting data, as well as calculating)
    * Often impossible

### Scenario 2
Administrative data

Some key features:
* Sampling fram contains a lot not in population
* Have access to entire frame

### Scenario 3
What we like to think we have

Key features:
* Optimistic sense that sampel is representative of population.

### Scenario 4
What we usually have

Key feature
* Sample may be drawn form a skewed frame and may not be representative of population

**Selection bias:** systematically facroting (or excluding) certain groups for inclusion in the sample  
**Non-response bias:** when people who don't respond are non-representative of the population

### Random Variables

* Empirical distribution: distribution of observed data (values and proportions)
* Probability distribution: a model for how the sample is generated. (values and probabilities)

A **random variable** is a variable that can take numerical values with particular probabilites i.e. $P(X=x)$

A function of random variables is *also* a random variable

### Bernoulli Distribution
A random variable that takes the value 1 with probability $p$ and 0 otherwise

$$P(X=1)=p$$
$$P(X=0)=1-p$$

### Binomial Distribution
A random variable that counts the number of "successes" in $n$ independent trials where each succeeds with probability $p$.

key word: ***independent***

$y$ is binomial($n, p$) if:

$$P(Y=y) = {n\choose y}p^y(1 - p)^{n - y}$$

### Types of Distributions

**Discrete**
* The set of possible values that $X$ can take on is either *finite* or *countably infinite*
* Values are separated by some fixed amount
* For instance, $X = 1, 2, 3, 4, \cdots$
* Examples of common distributions
    * Bernoulli
    * Binomial
    * Uniform

**Continuous**
* The set of possible values that $X$ can take on is *uncountable*.
* Typically, $X$ can be any real number in some interval (not just our counting numbers)
* Examples of common distributions
    * Uniform
    * Normal

### Expected Value
The **expected value** of a random variable X is the weighted average of the values of $X$, where the weights are the porbabilities of the values.

$$E(x) = \sum_{x_i}x_iP(X=x_i) = \mu$$

### Properties of Expected Values
Linear transformations:  
$$Z = aX + b$$
$$E(Z) = E(aX + b)$$
$$E(Z) = aE(X) + E(b)$$
$$E(Z) = aE(X) + b$$

Additivity:
$$W = X_1 + X_2$$
$$E(W) = E(X_1 + X_2)$$
$$E(W) = E(X_1) + E(X_2)$$

Linearity of Expectation:
$$V = aX_1 + bX_2$$
$$E(V) = E(aX_1 + bX_2)$$
$$E(V) = aE(X_1) + bE(X_2)$$

### Probability Mass Function
The probability mass function (PMF) or the distribution of a random variable $X$ provides the probability that $X$ takes on each of its possible values. Note that there are two essential rules:

$$\sum_{x \in X} P(X = x) = 1$$

$$\forall x \in X, 0 \leq P(X = x) \leq 1$$

The code below should output the PMF for a evenly probabalistic size sided die.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def plot_pmf(xs, probs, rv_name="X"):
    plt.plot(xs, probs, "ro", ms=12, mec="b", color="b")
    plt.vlines(xs, 0, probs, colors="b", lw=4)
    plt.xlabel("$x$")
    plt.ylabel("$P(X = x)$")
    plt.ylim(0, 1)
    plt.title("PMF of $X$");
```

```python
xk = np.arange(1, 7)
pk = (1 / 6, 1 / 6, 1 / 6, 1 / 6, 1 / 6, 1 / 6)

plot_pmf(np.arange(1, 7), np.repeat(1 / 6, 6))

plt.yticks(
    np.linspace(0, 1, 7),
    (
        "0",
        r"$\frac{1}{6}$",
        r"$\frac{2}{6}$",
        r"$\frac{3}{6}$",
        r"$\frac{4}{6}$",
        r"$\frac{5}{6}$",
        "1",
    ),
);
```
