## Data Sampling and Probability

### Terminologies
* Census counts everyone in a population
* Survey is a set of questions
* Sample is a subset of the population
    * They are often used to make inferences about the population
* Errors
    * Chance Error: random sample can vary from what is expected, in any direction
    * Bias: A systematic error in one direction
* Convenience sample is whoever you can get a hold of, they are not random
    * Not good for inference
    * Introduces bias 
* Quota sample
    * First specify desired subgroups and then reach those targets however you can
    * They are not random

### Big Idea of Sampling
* **population**: The group that you want to learn something about
* **sampling frame**: The list from which the sample is drawn
* **sample**: Who you actually end up sampling
    * A subset of your sampling frame

### Biases
* selection bias
    * Excluding or favoring particular groups
    * You can avoid by examining the sampling frame and method of sampling
* response bias
    * People don't respond truthfully
    * response bias is often seen as unavoidable 
    * We can avoid by examining the questions asked and method of surveying
* non-response bias
    * People don't always respond
    * We can avoid by keeping surveys short and persistent

### Random sampling
* random sample with replacement
* simple random sample (SRS)
    * Note that since this is without replacement, it will hurt for smaller sampling frames.

### Binomial and Multinomial Probabilities
* Binomial: $P(k \text{ successes}) = {n\choose{k}}p^k(1 - p)^{n - k}$
* Multinomial: $\frac{n!}{k_1!k_2!k_3!}p_1^{k_1}p_2^{k_2}p_3^{k_3}$

## Estimation and Bias
Statistical Bias:  
*The difference between your estimate and the truth*

Types of Distributions:  
In Data 100, we'll be focusing on two types of distributions:
* Bernoulli Distribution
    * A random variable that takes the value 1 with probability $p$ and 0 otherwise
    * $X \sim Bernoulli(p)$ where $P(X=1) = p, P(X=0)=1-p$
* Binomial Distribution
    * A random variable that counts the number of successes in $n$ independent trials where each succeeds with probability $p$
    * $X \sim Binomial(n, p)$ where $P(Y=y) = {n\choose{y}}p^y(1 - p)^{n - y}

Recall that there are two different types of distributions: Continuous and Discrete. Let's categorize them to see the bigger picture:
* Discrete
    * Bernoulli
    * Binomial
    * Uniform or finite set
* Continuous
    * Uniform on the unit interval
    * Normal($\mu, \sigma^2$)

Expected value of a random variable $X$ is the weighted average of the values of $X$, where the weights are the probabilites of the values. (Analagous to average)
* $E[X] = \sum_{x_i}x_iP[X = x_i] = \mu$

The `SQL` Query Flow
```sql
SELECT [DISTINCT] <column expression list> 
FROM <list of tables> 
WHERE <predicate>
GROUP BY <column list>
HAVING <predicate>
ORDER BY <column list> 
LIMIT <number of rows>;
```
And the ordering for how each query is processed goes like this:
`FROM -> WHERE -> GROUP BY -> HAVING -> SELECT -> ORDER BY -> LIMIT`

Here are functions and methods to consider:

    * `df[['name', ... , 'name']]` will output a Dataframe

## Data Cleaning and EDA
Here lies the big idea of data science, where we have an infinite loop of data cleaning and exploratory data analysis (EDA)

*Getting to know the data*

The process of transforming, visualizing, and summarizing data to
* build/confirm understanding of data
* identify and address potential issues in the data
* inform the subsequent analysis
* discover potential hypothesis

EDA is an open ended analysis:
Always be willing to find something surprising, thus this is why it's an infinite cycle

Here are the big ideas of EDA
* Structure
* Granularity
* Scope
* Temporality
* Faithfulness

Here are things to consider about the data, where we split it up into 4 subcategories!
* Quantitative
    * Continuous

        * e.g. number of siblings, yrs of education
* Qualitative
    * Ordinal
        * e.g. Preferences, level of education
