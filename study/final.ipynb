* A random variable is a numerical function of a random sample.
* Expectation can be seen as the weighted average.
    * Popular formulation is $$E(X) = \sum_x xP(X=x)$$
* Note that linearity can apply to expecation, such that we can have a formulation as such: (linearity of expectation) $$E(aX + bY) = aE(X) + bE(Y)$$
* Variance is the expected squared deviation form the expectation of X: $$Var(X) = E((X - E(X))^2)$$
* Standard deviation is just the square root of variance: $$SD(X) = \sqrt{Var(X)}$$
    * Variance and standard deviation are both non negative!
* Important derivation of variance:
$$Var(X) = E((X - E(X))^2)$$
$$Var(X) = E(X^2 - 2XE(X) + (E(X)^2)$$
$$Var(X) = E(X^2) - 2XE(X)E(X) + (E(X))^2$$
$$Var(X) = E(X^2) - (E(X))^2$$

There is also the covariance:
* Variance is only additive when the covariance is 0: $$Var(X + Y) = Var(X) + Var(Y) + 2Cov(X, Y)$$
* Correlation is denoted as $r$ and is usually represented as: $$r(X, Y) = \frac{Cov(X, Y)}{SD(X)SD(Y)}$$
And
$$Cov(X, Y) = 0$$
$$r(X, Y) = 0$$
are both true at the same time, or not.
Covariance = 0 means that the variables are uncorrelated. For example, independent random variables are uncorrelated.

## Gradient Descent Algorithm
* $\alpha$ is the learning rate, treat it like a hyper parameter, the large it is, the longer the step. The smaller it is, the longer it takes to converge to the optimum.
* $$x^{(t + 1)} = x^{(t)} - \alpha\frac{d}{dx}f(x)$$
* Only works for convex functions, as gradient descent potentially fails for multiple minima.
* Convex just means only one local minima. 
* $$tf(a) + (1 - t)f(b) \geq f(ta + (1 - t)b) \forall a, b \in f, t ]in[0, 1]$$
* stochastic gradient descent is much more noisier since it takes a batch/subsamble from the training set.

## PCA/SVD (Unsupervised Learning)
* In PCA, we often want to reduce the dimensionality to better generalize data. It is a great preprocessing method. 
* Consider the dimensionality of the data/design matrix as the rank of the matrix!
* SVD can decompose a matrix into $U\Sigma V^T$
    * $U\Sigma$ is the rescaled data and $V^T$ is the transormation matrix.
    * Recall that $U$ and $V$ are orthonomal matrices (orthonormal columns). 
    * $\Sigma$ is a diagonla matrix containing the singular values of original design matrix $X$
