## Terminology
In a relational database, each table is a relation,
and in that row is a record or tuple
and in that is a column which is also called an attribute or field.
Arritbutes have names, while attributes also have data types
attributes also have constraints
Attributes may also be marked as primary or foreign keys

Primary key just invokes that a row or tuple needs to be unique.

## Syntax
Here is just a generic SQL Syntax
```sql
SELECT [DISTINCT] <column>
FROM <table>
WHERE <predicate>
GROUP BY <column>
HAVING <predicate>
ORDER BY <column>
LIMIT <number>
```

## Joins in SQL

### Cross Joins
Join everything, so if you have 2 tables with 4 rows, output will be 16 rows.

### Inner Join

Conceptually, an inner join is a cross join followed by removal of bad rows.

## NULL Values
Field values are sometimes unknown, and this is represented as `NULL` in SQL.
Null also introduces new issues in selections with WHERE adn aggregatin.
Expect NULLS when performing queries such as outer joins.

Sometimes it's helpful to do explicit null checks such as:
```sql 
SELECT * FROM student WHERE name IS NULL;
```

## SQL Predicates
In addition to numerical comparisons, there are also built in predicates!

You can also use casting to cast types (such as ints, floats, etc)

You can also simulate sampling by using the clause `LIMIT`

A random sampling could be done like this (varies upon database engine):
```sql
SELECT * 
FROM action_movie 
ORDER BY RANDOM() 
LIMIT 3
```

## Common Table Expressions (CTE)
Allows for temporary tables to help organiz more complex queries.
* can be used instead of subqueries
* make comple queries more readable

Ex.
```sql
WITH t2 AS (
SELECT * FROM t ORDER BY RANDOM() LIMIT 10
)
SELECT * FROM t2, u;
```
